### Exercise 3: Training a Classifier

Lets' repeat what we did manually in the previous lab using scikit-learn. We'll use a LDA classifier, which we can instanciate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get a unbias assessment of the performance of Machine Learning algorithms)

for example, here we'll take use 4M events for training and the remainder for testing.

```python
N_Train=400000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]
```

We can train the classifier as follow:

```python
Fisher.fit(X_Train,y_Train)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare ROC curves computed on the test versus training samples, in a single plot. Do you see a bias?

```python
from sklearn.metrics import roc_curve, roc_auc_score
```

```python
train_scores = Fisher.decision_function(X_Train)
test_scores = Fisher.decision_function(X_Test)

fpr_train, tpr_train, _ = roc_curve(y_Train, train_scores)
auc_train = roc_auc_score(y_Train, train_scores)

fpr_test, tpr_test, _ = roc_curve(y_Test, test_scores)
auc_test = roc_auc_score(y_Test, test_scores)

plt.figure(figsize=(8,6))
plt.plot(fpr_train, tpr_train, label=f"Train ROC (AUC = {auc_train:.3f})", color="red", linestyle="--")
plt.plot(fpr_test, tpr_test, label=f"Test ROC (AUC = {auc_test:.3f})", color="blue")
plt.plot([0, 1], [0, 1], 'k--', label="Random Guess")

plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.title("ROC Curve Comparison: Training vs Test")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

#### Part b

Train the Fisher performance of using the raw, features, and raw+features as input. Compare the performance one a single plot. 

```python
RawNames = ['l_1_pT', 'l_1_eta', 'l_1_phi',
            'l_2_pT', 'l_2_eta', 'l_2_phi',
            'MET', 'MET_phi']

FeatureNames = ['M_R', 'MET_rel', 'R', 'M_TR_2', 'MT2',
                'S_R', 'M_Delta_R', 'dPhi_r_b', 'cos_theta_r1', 'axial_MET']

AllFeatures = RawNames + FeatureNames
```

```python
def fisher_roc(features, X_train_df, X_test_df, y_train, y_test):
    clf = DA.LinearDiscriminantAnalysis()
    clf.fit(X_train_df[features], y_train)

    scores_train = clf.decision_function(X_train_df[features])
    scores_test = clf.decision_function(X_test_df[features])

    fpr, tpr, _ = roc_curve(y_test, scores_test)
    auc_score = roc_auc_score(y_test, scores_test)

    return fpr, tpr, auc_score

fpr_raw, tpr_raw, auc_raw = fisher_roc(RawNames, Train_Sample, Test_Sample, y_Train, y_Test)
fpr_feat, tpr_feat, auc_feat = fisher_roc(FeatureNames, Train_Sample, Test_Sample, y_Train, y_Test)
fpr_all, tpr_all, auc_all = fisher_roc(AllFeatures, Train_Sample, Test_Sample, y_Train, y_Test)
```

```python
plt.figure(figsize=(8,6))
plt.plot(fpr_raw, tpr_raw, label=f"Raw Features (AUC = {auc_raw:.3f})", color="blue")
plt.plot(fpr_feat, tpr_feat, label=f"High-Level Features (AUC = {auc_feat:.3f})", color="green")
plt.plot(fpr_all, tpr_all, label=f"Combined Features (AUC = {auc_all:.3f})", color="red")
plt.plot([0,1], [0,1], 'k--', label="Random Guess")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Fisher Discriminant ROC Comparison")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
##Part a
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
```

```python
from sklearn.metrics import roc_curve, auc

def compare_classifier_inputs(clf, clf_name, df, RawNames, FeatureNames, N_Train=400000):
    # Split data
    Train_Sample = df[:N_Train]
    Test_Sample = df[N_Train:]
    
    y_Train = Train_Sample["signal"]
    y_Test = Test_Sample["signal"]
    
    input_groups = {
        "Raw": RawNames,
        "Features": FeatureNames,
        "Raw + Features": RawNames + FeatureNames
    }
    
    plt.figure(figsize=(10, 7))
    
    for label, cols in input_groups.items():
        X_Train = Train_Sample[cols]
        X_Test = Test_Sample[cols]
        
        clf.fit(X_Train, y_Train)
        
        # Decision score or probability
        if hasattr(clf, "decision_function"):
            y_scores = clf.decision_function(X_Test)
        else:
            y_scores = clf.predict_proba(X_Test)[:,1]
        
        fpr, tpr, _ = roc_curve(y_Test, y_scores)
        auc_score = auc(fpr, tpr)
        
        plt.plot(fpr, tpr, label=f"{label} (AUC = {auc_score:.3f})")

    plt.plot([0, 1], [0, 1], 'k--')
    plt.title(f"Classifier Performance: {clf_name}")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
Log_Reg = LogisticRegression(max_iter=1000)
```

```python
compare_classifier_inputs(Log_Reg, "Logistic Regression", df, RawNames, FeatureNames)
```

```python
RF = RandomForestClassifier(n_estimators=100)
```

```python
compare_classifier_inputs(RF, "Random Forest", df, RawNames, FeatureNames)
```
