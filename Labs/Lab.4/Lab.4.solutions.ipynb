```python
class Counter:
    def __init__(self,max_num,count=0,step=1):
        self.__max_num = max_num
        self.__count = count ##Should someone want to they don't have to start at zero
        self.__step = step ##should have some base increment step
    
    def increment(self):
        if self.__count + self.__step > self.__max_num:
            return "Cannot go past max"
        else:
            self.__count += self.__step
            
    def reset_count(self):
        self.__count = 0
        print("Counter has been reset to zero.")
        
    def get_count(self):
        return self.__count
    
    def get_max(self):
        return f"The maximum number to increment to is {self.__max_num}"
    
    def is_at_max(self):
        if self.__max_num <= self.__count:
            print("Count is past/at maximum value.")
        else:
            diff = self.__max_num - self.__count
            print(f"Counter is {diff} units away.")
```

```python
class foo:
    def __init__(self,a,b=None):
        self.a=a
        self.b=b
        
    def __repr__(self):
        return "foo("+repr(self.a)+","+repr(self.b)+")"
    
    def save(self,filename):
        f=open(filename,"w")
        f.write(self.__repr__())
        f.close()
        
   
def foo_loader(filename):
    f=open(filename,"r")
    tmp=eval(f.read())
    f.close()
    return tmp
```

```python
# Test
print(repr(foo(1,"hello")))
```

```python
# Create an object and save it
ff=foo(1,"hello")
ff.save("Test.foo")
```

```python
# Load the object
ff_reloaded=foo_loader("Test.foo")
ff_reloaded
```
