```python
def compare_scatter(df,x_var_name,y_var_name,selections,make_legend=True) :
    for label,selection in selections.items():   
        _=plt.scatter(df[selection][x_var_name],df[selection][y_var_name],label=label)

    _=plt.xlabel(x_var_name)
    _=plt.ylabel(y_var_name)

    if make_legend:
        _=plt.legend()
```

```python
import matplotlib.pyplot as plt

def pair_plot_group(df, columns, selections, figsize=(15, 15), bins=50, alpha=0.5, density=True):
    """
    Create pair plots for the provided columns using given selections.
    
    Parameters:
    - df: pandas DataFrame containing the data.
    - columns: list of column names to include in the pair plot.
    - selections: dictionary with keys as labels and values as boolean selections for df.
    - figsize: size of the overall figure.
    - bins: number of bins for histogram plots (diagonal).
    - alpha: transparency for histograms.
    - density: whether to normalize the histograms.
    """
    
    n_columns = len(columns)
    plt.figure(figsize=figsize)
    
    plot_i = 0
    for i, x_var_name in enumerate(columns):
        for j, y_var_name in enumerate(columns):
            plot_i += 1
            plt.subplot(n_columns, n_columns, plot_i)
            
            # Only show the legend in the first subplot
            make_legend = (plot_i == 1)
            
            if i == j:
                # On the diagonal: plot histograms of the variable
                for label, selection in selections.items():
                    plt.hist(df[selection][x_var_name],
                             label=label,
                             alpha=alpha,
                             density=density,
                             bins=bins)
                if make_legend:
                    plt.legend()
            else:
                # Off-diagonals: plot scatter plots for pairs of variables
                for label, selection in selections.items():
                    plt.scatter(df[selection][x_var_name],
                                df[selection][y_var_name],
                                label=label)
                plt.xlabel(x_var_name)
                plt.ylabel(y_var_name)
                if make_legend:
                    plt.legend()
                    
    plt.tight_layout()
    plt.show()

```

```python
pair_plot_group(df, RawNames, selection_dict, figsize=(12, 12))
```

```python
pair_plot_group(df, FeatureNames, selection_dict, figsize=(12, 12))
```
